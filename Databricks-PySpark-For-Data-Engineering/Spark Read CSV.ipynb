```python
# Read data file(CSV) with no header

schema_str = "id INT, customer_name STRING, product STRING, quantity INT, price double, remarks STRING"

df = spark.read\
.schema(schema_str)\
.csv("/Volumes/my_files/my_schema/csv_files/my_csv_files/sales_no_header.csv")


display(df)
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType

schema = StructType([
    StructField("id", StringType(), True),
    StructField("customer_name", StringType(), True),
    StructField("my_product", StringType(), True),
    StructField("quantity", IntegerType(), True),
    StructField("price", DoubleType(), True),
    StructField("remarks", StringType(), True)
])

df=spark.read.\
schema(schema)\
.csv("/Volumes/my_files/my_schema/csv_files/my_csv_files/sales_no_header.csv")

display(df)
```

```python
# Read data file(CSV) with pipe as delimiter

df = spark.read\
.option("header", "true")\
.option("inferSchema", "true")\
.option("delimiter", "|")\
.csv("/Volumes/my_files/my_schema/csv_files/my_csv_files/sales_pipe.csv")


display(df)
```

```python
# Read data files(csv) with quote characters

df = spark.read\
.option("header", "true")\
.option("inferSchema", "true")\
.option("delimiter", "|")\
.option("quote", "'")\
.csv("/Volumes/my_files/my_schema/csv_files/my_csv_files/sales_pipe_single_quote.csv")


display(df)
```

```python
# Read data files(csv) with single quote without escape character

df = spark.read\
.option("header", "true")\
.option("inferSchema", "true")\
.option("delimiter", "|")\
.option("quote", "'")\
.csv("/Volumes/my_files/my_schema/csv_files/my_csv_files/sales_pipe_single_quote_escape.csv")


display(df)
```

```python

# Read data files(csv) with single quote with escape character
df = spark.read\
.option("header", "true")\
.option("inferSchema", "true")\
.option("delimiter", "|")\
.option("quote", "'")\
.option("escape", "#")\
.csv("/Volumes/my_files/my_schema/csv_files/my_csv_files/sales_pipe_single_quote_escape.csv")


display(df)
```

```python
# Read data files(csv) with single quote with escape character in multiline

df = spark.read\
.option("header", "true")\
.option("inferSchema", "true")\
.option("delimiter", "|")\
.option("quote", "'")\
.option("escape", "#")\
.option("multiline", "true")\
.csv("/Volumes/my_files/my_schema/csv_files/my_csv_files/sales_pipe_single_quote_escape_multiline.csv")


display(df)

```
