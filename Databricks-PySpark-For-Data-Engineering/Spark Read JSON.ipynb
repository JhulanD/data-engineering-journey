```python
# Single line JSON file read in Spark
df = spark.read.json("/Volumes/my_files/my_schema/sample_datasets/json_files_uploaded/sales_single_line.json")

display(df)
```

```python
# Multiline JSON file read in Spark
df = spark.read\
.option("multiline", "true")\
.json("/Volumes/my_files/my_schema/sample_datasets/json_files_uploaded/sales_multi_line.json")


display(df)
```

```python
# Write JSON file in Spark
df.write.format("json")\
.save("/Volumes/my_files/my_schema/sample_datasets/json_out/json_out_01")
```

```python
# Write CSV file in Spark
df.write.format("csv")\
.save("/Volumes/my_files/my_schema/sample_datasets/csv_out/csv_out_01")
```
