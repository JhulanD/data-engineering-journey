**Extract Divisibles:**

*Given numbers = [10, 15, 20, 25, 30, 35, 40], use a list comprehension to create a new list containing only the numbers that are divisible by 5 and greater than 15.*

```python
numbers = [10, 15, 20, 25, 30, 35, 40]

new_numbers = [num for num in numbers if (num % 5)==0 and num > 15]

print(new_numbers)
```

**Uppercase Long Words:**

*Given words = ['apple', 'cat', 'banana', 'dog', 'elephant'], write a list comprehension that converts each word to uppercase, but only for words with more than 3 letters.*

```python
words = ['apple', 'cat', 'banana', 'dog', 'elephant']

words_uppercase = [word.upper() for word in words if len(word) > 3]

print(words_uppercase)
```

**Number-Square Tuples:**

*Given nums = [1, 2, 3, 4, 5], use a list comprehension to generate a list of tuples where each tuple contains the original number and its square, formatted as (number, square)*

```python
nums = [1, 2, 3, 4, 5]

num_square = [(num, num * num) for num in nums]

print(num_square)
```

In Python, you can assign a **default value to a function parameter** by using the assignment operator (=) in the function definition. If a caller omits that argument, **Python automatically falls back to the preset default**.

```python
def greet(name, greeting = 'Hello'):
    print(f"{greeting}, {name}!")
```

```python
# Uses the default value for 'greeting'

greet('Alice')
```

```python
# Overrides the default value

greet('Bob', 'Goodbye')
```

### ⚠️ The Golden Rule: Required Before Optional
***All parameters without default values must be placed before parameters with default values. Mixing up this order will trigger a SyntaxError.***


```python
def set_profile(status = 'Active', username):
    pass
```

```python
def set_profile(username, status = "Active"):
    pass
```

***🚨 Crucial Trap: Mutable Default Arguments***


```python
# In Python, default arguments are computed exactly once when the function is first defined,
# not every time the function is run. If you use a mutable object like a list or a dictionary as a default value,
# the exact same object is shared and reused across every single function call

def add_item(item, item_list=[]):
    item_list.append(item)
    return item_list

print(add_item("apple"))  # Output: ['apple']
print(add_item("banana")) # Output: ['apple', 'banana']  (The list persists!)


```

```python
# The Pythonic Fix
# To fix this, use None as a placeholder (sentinel) and initialize a new list inside the function:


def add_item(item, item_list = None):
    if item_list is None:
        item_list = [] #Creates a barnd new list each time
        item_list.append(item)
        return item_list


print(add_item('apple'))
print(add_item('banana'))
```

### **Python *args and **kwargs


```python
# Keyword Arguments (**kwargs)
# The **kwargs syntax collects extra named keyword arguments into a dictionary.

def sum_numbers(*args):
    # args is a tuple containing all passed positional arguments
    return sum(args)

print(sum_numbers(1,2,4,6,8))
```

```python
# Keyword Arguments (**kwargs)
# The **kwargs syntax collects extra named keyword arguments into a dictionary

def building_info(**kwargs):
    # **kwargs is a dictionary of key-value pairs
    for key, value in kwargs.items():
        print(f"{key}: {value}")

building_info(City = 'Kolkata', State = 'WB', Country = 'India')
```
