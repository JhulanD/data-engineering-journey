### 🔹 What are Variables?
In Python, variables are named containers or labels that point to data values stored in your computer's memory. 
They make code reusable, clean, and less redundant.
Unlike other programming languages, you do not need to explicitly declare a variable or its type before using it; Python automatically creates the variable the exact moment you assign a value to it.


**Rules for naming variables:**
1. Names cannot start with a number.  
2. No spaces — use `_` instead.  
3. Avoid special characters (`!@#$%^&*`).  
4. Case-sensitive (`Data` ≠ `data`).  

**Syntax:**

```python
# 🔹 Examples: Declaring & Using Variables

# Integer variable
records = 1000

# Float variable
storage_cost_per_gb = 0.12

# String variable
pipeline_name = "Daily_ETL_Pipeline"

# Boolean variable
is_active = True

# Printing variables
print(records) 
print(storage_cost_per_gb)
print(pipeline_name)
print(is_active)
```

```python
a = 42        # int
b = 32.30     # float
c = True      # bool

print(type(a))   # <class 'int'>
print(type(b))   # <class 'float'>
print(type(c))   # <class 'bool'>
```

```python
# -------------------------------------------------------------
# ARITHMETIC OPERATIONS IN PYTHON
# -------------------------------------------------------------

# Define our initial variables
a = 10
b = 3

# 1. Addition (+)
# Adds two values together
addition_result = a + b
print(f"Addition: {a} + {b} = {addition_result}")  # Output: 13


# 2. Subtraction (-)
# Subtracts the right value from the left value
subtraction_result = a - b
print(f"Subtraction: {a} - {b} = {subtraction_result}")  # Output: 7


# 3. Multiplication (*)
# Multiplies two values
multiplication_result = a * b
print(f"Multiplication: {a} * {b} = {multiplication_result}")  # Output: 30


# 4. True Division (/)
# Divides the left value by the right value.
# Note: This ALWAYS returns a float, even if the result is a whole number.
division_result = a / b
print(f"True Division: {a} / {b} = {division_result}")  # Output: 3.3333333333333335
print(f"Type of true division output: {type(division_result)}")  # Output: <class 'float'>


# 5. Floor Division (//)
# Divides the numbers and rounds DOWN to the nearest whole integer.
# It chops off everything after the decimal point.
floor_division_result = a // b
print(f"Floor Division: {a} // {b} = {floor_division_result}")  # Output: 3


# 6. Modulus (%)
# Divides the left value by the right value and returns only the remainder.
# Great for checking if a number is even (num % 2 == 0)
modulus_result = a % b
print(f"Modulus (Remainder): {a} % {b} = {modulus_result}")  # Output: 1


# 7. Exponentiation (**)
# Raises the left number to the power of the right number (a squared/cubed etc.)
exponent_result = a ** b
print(f"Exponentiation: {a} ** {b} = {exponent_result}")  # Output: 1000 (10 to the power of 3)


# -------------------------------------------------------------
# BONUS: Operator Precedence (PEMDAS / BODMAS)
# Python follows standard mathematical rules for the order of operations.
# -------------------------------------------------------------
mixed_calculation = 5 + 3 * 2 ** 2  # Exponent first (4), then Multiply (12), then Add (17)
print(f"Mixed Calculation: {mixed_calculation}")  # Output: 17

# Use parentheses () to force Python to calculate specific parts first
grouped_calculation = (5 + 3) * 2  # Add first (8), then Multiply (16)
print(f"Grouped Calculation: {grouped_calculation}")  # Output: 16
```

#### Python Variables: Syntax, Usage, and Examples

Python variables are containers with names that store values like numbers, strings, and lists. In essence, variables make it easier to access and modify the values they store as a Python application runs.

##### Quick Answer: How to Use Variables in Python

In Python, you create a variable by assigning a value to a name using the equals sign (=). Unlike some languages that use a $ prefix *(like PHP or PowerShell)*, Python variables are just plain names.

Python is **dynamically typed**, meaning you don't need to declare the variable's type; it's determined automatically from the assigned value.

```python
# Assigning a string value to a variable
greeting = "Hello, World!"

# Assigning a number value to a variable
user_age = 30

# Using the variables
print(greeting)
# Outputs: Hello, World!

print(f"The user is {user_age} years old.")
# Outputs: The user is 30 years old.
```

```python
import math

"""
=====================================================================
The 'math' module is a built-in Python library that 
provides access to essential mathematical constants and functions 
defined by the C standard. It operates entirely on real numbers 
and always returns floating-point values for operations that can 
result in decimals. It does not support complex (imaginary) numbers.
=====================================================================
"""

# =====================================================================
# 1. MATHEMATICAL CONSTANTS
# =====================================================================
print("--- Constants ---")

# math.pi represents the ratio of a circle's circumference to its diameter.
# Value: Approximately 3.141592653589793
print(f"Pi (π): {math.pi}")

# math.e represents Euler's number, the base of natural logarithms.
# Value: Approximately 2.718281828459045
print(f"Euler's number (e): {math.e}")

# math.tau represents the circle constant equal to 2*pi (radii per circumference).
# Value: Approximately 6.283185307179586
print(f"Tau (τ): {math.tau}")


# =====================================================================
# 2. ROUNDING AND NUMBER ALTERATION
# =====================================================================
print("\n--- Rounding & Number Functions ---")
value = 4.2
negative_value = -4.8

# math.ceil(x) rounds a number UP to the nearest integer.
# Example: 4.2 becomes 5.
print(f"Ceil (up): {math.ceil(value)}")

# math.floor(x) rounds a number DOWN to the nearest integer.
# Example: 4.2 becomes 4.
print(f"Floor (down): {math.floor(value)}")

# math.trunc(x) cuts off the decimal part entirely, leaving just the integer.
# Unlike floor(), it moves toward zero for negative numbers. -4.8 becomes -4.
print(f"Truncate: {math.trunc(negative_value)}")

# math.fabs(x) returns the absolute value of a number, forcing it to be positive.
# It always returns the result as a floating-point number.
print(f"Absolute Value: {math.fabs(negative_value)}")


# =====================================================================
# 3. POWERS, ROOTS, AND LOGARITHMS
# =====================================================================
print("\n--- Powers, Roots, & Logarithms ---")

# math.sqrt(x) calculates the square root of a number. Returns a float.
# Note: Passing a negative number here will cause a ValueError.
print(f"Square Root of 16: {math.sqrt(16)}")

# math.pow(x, y) raises x to the power of y (x^y). It always returns a float.
print(f"2 raised to power 3: {math.pow(2, 3)}")

# math.log(x, [base]) calculates the logarithm of x to the given base.
# If you omit the base argument, it defaults to the natural log (base e).
print(f"Natural Log of e: {math.log(math.e)}")
print(f"Log base 10 of 100: {math.log(100, 10)}")


# =====================================================================
# 4. TRIGONOMETRY AND ANGLE CONVERSIONS
# =====================================================================
print("\n--- Trigonometry ---")
degrees = 90

# Python's sin(), cos(), and tan() functions require angles to be in radians.
# math.radians(x) converts an angle from degrees to radians.
radians = math.radians(degrees)
print(f"90 degrees in radians: {radians}")

# math.sin(x) returns the sine of an angle given in radians.
print(f"Sine of 90 degrees: {math.sin(radians)}")

# math.cos(x) returns the cosine of an angle given in radians.
# Due to tiny computer floating-point limitations, this returns close to 0 (e.g., 6.12e-17).
print(f"Cosine of 90 degrees: {math.cos(radians)}")

# math.degrees(x) converts an angle from radians back into degrees.
print(f"Convert back to degrees: {math.degrees(radians)}")


# =====================================================================
# 5. ADVANCED ARITHMETIC & COMBINATORICS
# =====================================================================
print("\n--- Combinatorics & Special Functions ---")

# math.factorial(x) returns the factorial of an integer (e.g., 5! = 5 * 4 * 3 * 2 * 1).
# Note: This function only accepts positive whole numbers.
print(f"Factorial of 5 (5!): {math.factorial(5)}")

# math.comb(n, k) calculates the number of ways to choose k items from n items without repetition.
# Also known as "n choose k" or combinations.
print(f"Combinations (5 choose 2): {math.comb(5, 2)}")

# math.gcd(*integers) returns the Greatest Common Divisor of the provided numbers.
# It finds the largest whole number that divides into both values without a remainder.
print(f"Greatest Common Divisor of 24 & 36: {math.gcd(24, 36)}")

# math.hypot(*coordinates) calculates the Euclidean norm, sqrt(x² + y²).
# This is ideal for finding the hypotenuse of a right-angled triangle.
print(f"Hypotenuse of right triangle (3, 4): {math.hypot(3, 4)}")


```

### 80/20 Data Engineering Math Guide

According to the Pareto Principle (80/20 rule), **20% of math concepts handle 80% of data engineering tasks**.

In production pipelines, you rarely use advanced trigonometry. Instead, you focus heavily on data cleaning, scaling, and batch allocation.

---

### 1. Data Quality Checks (The Absolute Essentials)
Production data from APIs, databases, and CSVs is often messy. Use these functions to filter out bad rows before loading data into your target warehouse.

* **`math.isnan(x)`**: Checks for missing or corrupt values (`NaN`).
* **`math.isinf(x)`**: Catches pipeline math errors, such as dividing a metric by zero.

---

### 2. Batching and Partitioning (Data Chunking)
Data engineers move massive datasets in smaller chunks (batches) to optimize memory usage and prevent system crashes.

* **`math.ceil(x)`**: Rounds **up**. If you have 105 rows and a batch size of 10, `math.ceil(105/10)` ensures your pipeline schedules **11 batches** so zero data is left behind.
* **`math.floor(x)`**: Rounds **down**. Useful for strict partitioning limits and allocating data into fixed-size buckets.

---

### 3. Feature Scaling (Log Transformations)
Extreme outliers (e.g., a few multi-millionaires in an income dataset) skew analytics. Log transformations compress these ranges for better downstream processing.

* **`math.log1p(x)`**: Calculates $\log(1 + x)$. This is the standard tool for normalization because it safely processes `0` values without crashing your pipeline.

---

### ⚠️ The 80/20 Catch: Scaling Up
The built-in Python `math` module only processes **one number at a time** (scalar execution). 

In production data engineering, you will apply this exact same logical 20% using vectorised libraries like **NumPy** or **Pandas** to process billions of rows simultaneously.

| Python `math` (Single Value) | NumPy / Pandas (Millions of Rows) |
| :--- | :--- |
| `math.isnan(val)` | `np.isnan(array)` or `df.isna()` |
| `math.ceil(val)` | `np.ceil(array)` |
| `math.log1p(val)` | `np.log1p(array)` |

```python
# Method 1: String Slicing (Recommended)
s = 'Hello'
s = 'J' + s[1:]
print(s)  # Output: Jello
```

```python
# Method 2: The replace() Method
s = 'Hello'
s = s.replace('H', 'J', 1)  # The '1' ensures only the first 'H' is replaced
print(s)  # Output: Jello
```

* In Python, strings are immutable, which means their values cannot be changed after they are created.

* The method name.upper() creates and returns a completely new string, but it does not modify the original name variable.

** If you want to update the variable to uppercase, you must reassign the output back to it:

```python
name = 'data'
name = name.upper()  # Reassigns the new uppercase string to name
print(name)          # Output: DATA
```

```python
# int() truncates — it simply cuts off the decimal part, so 3.9 becomes 3.
# The answer 4 assumes rounding, which is the most common misconception here;
# if you actually want nearest-integer behavior you must call round(3.9) instead.

print(int(3.9))
```

```python
# The / operator always returns a float in Python 3, even when the division is exact,
# so 10 / 5 prints 2.0. The integer 2 is what you would get from floor division 10 // 5 
# mixing up which operator preserves int results is a frequent source of subtle dtype bugs in data code.

print(10 / 5)
```

```python
# "append" adds its argument as a single item, so the whole list [3, 4] lands inside l as one nested element.
# The flat [1, 2, 3, 4] is what "extend" would produce — extend unpacks the items and adds them individually,
# and confusing the two is one of the most common beginner list bugs.

l = [1, 2]
l.append([3, 4])
print(l)
```

```python
# Sets are unordered — they do not track positions at all, so s[0] raises a TypeError.
# 'Prints 10' is tempting because 10 was written first, but insertion order is not preserved or addressable in a set;
# if you need positional access, you want a list, and if you only need membership and uniqueness, a set is right.

s = {10, 20, 30}
print(s[0])
```
